```python
import numpy as np
import time
import matplotlib.pyplot as plt
import matplotlib.colors as colors
from joblib import Parallel, delayed
import spintoolkit_py as sptk
```

```python
# Define units
unit0 = sptk.units(energy="meV", length="angstrom")
```

```python
# the position of sublattice atoms in fractional coordinates, in the unit of a basis vector
pos_sub = np.array([
    				[0.5, 0.0, 0.0],
    				[0.0, 0.5, 0.0],
    				[0.5, 0.0, 0.5],
    				[0.0, 0.5, 0.5]
					])
# the length unit is angstrom
# 2D lattice vectors
a = np.array([
    [5.326 * unit0.angstrom, 0.0, 0.0],
    [0.0, 5.602 * unit0.angstrom, 0.0],
    [0.0, 0.0, 7.635 * unit0.angstrom],
])

# superlattice A = 1 x 1
A = np.array([1, 1, 1])
```

```python
# define the lattice
latt = sptk.lattice(basis_a=a, pos_sub=pos_sub, l = A)
hamiltonian = sptk.model_spin(S=2.5, mode="dipole", lattice=latt)
total_sites = latt.total_sites()
print()
```

```python
A = 1.0
B = 1.0
C = 1.0
D = 1.0
E = 1.0
F = 1.0
H = 0.01	# use H for external magnetic field
g = 2.0

```

```python
# our sites : 0, 1, 2 ,3

for site_i in range(total_sites):
	coor_i, sub_i = latt.site2coor(site=site_i)
	coor0_i, r̃i = latt.r2superlattice(coor=coor_i)

	# To add Allowed exchange matrix: [A  D -F
	#                                  D  B  E
	#                                  F -E  C]
	#                                  on Bond(0, 2, [0, 0, 0])
	if sub_i == 0:
		sub_j = 2
		coor_j = [coor_i[0] + 0, coor_i[1] + 0, coor_i[2] + 0]	# [0, 0,]
		site_j = latt.coor2site(coor=coor_j, sub=sub_j)
		coor0_j, r̃j = latt.r2superlattice(coor=coor_j)

		Jmatrix = sptk.Mat3(A, D, -F,
           					D, B, E,
           					F, -E, C)

		hamiltonian.add_2spin_Jmatrix(Jij = Jmatrix,
                              		  site_i = site_i, site_j = site_j,
                              		  rtilde_i = r̃i, rtilde_j = r̃j
									  )
```
